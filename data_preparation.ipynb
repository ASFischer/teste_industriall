```python
#Para rodar os próximos comandos deve primeiro descompactar a pasta new_data encontrada no arquivo dados.zip

import zipfile

arquivo_zip = 'dados.zip'

with zipfile.ZipFile(arquivo_zip, 'r') as zip_ref:
    zip_ref.extractall('.')
```

```python
import pandas as pd
import glob
import matplotlib.pyplot as plt

#Busca por todos arquivos que seguem o padrão descrito. O (*) pega desde o arquivo 00 até o 51
feature_files = glob.glob('new_data/TAG_iALL_PS_*.csv')

#Defino a lista que guardará as tabelas
list_dfs = []

#Loop para ler os arquivos e já transformar o timestamp em data e definir como índice
for files in feature_files:
    temp_df = pd.read_csv(files, index_col='timestamp', parse_dates=['timestamp'])
    list_dfs.append(temp_df)

feature_df = pd.concat(list_dfs, axis=1)
#Simplificando as variáveis para somente números
feature_df.columns = feature_df.columns.str.replace('TAG_iALL_PS_', '')
```

```python
#Lendo e garantindo o index correto da variável target
target_df = pd.read_csv('new_data/target_iALL_PS.csv', index_col='timestamp', parse_dates=['timestamp'])

#Junção dos dois df
#'inner' vai garantir que somente as linhas(timestamp) presentes em ambas tabelas serão mantidas
final_df = feature_df.join(target_df, how='inner')

import os
if not os.path.exists('final_data'):
    os.makedirs('final_data')

final_df.to_csv('final_data/final_data.csv', index=False)
```
