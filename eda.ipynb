```python
import pandas as pd
import missingno as msno
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('final_data/final_data.csv')
print(df.shape)
df.head()
```

```python
df['target_iALL_PS'].value_counts()
```

```python
#Transformando a variável target para formato binário (0 = NORMAL, 1 = ANORMAL)
df['binary_target'] = df['target_iALL_PS'].apply(lambda x: 0 if x == 'NORMAL' else 1)
df.drop(columns=['target_iALL_PS'], inplace=True)
df.head()
```

```python
#Checando a quantidade de falhas
print(df['binary_target'].value_counts())
sns.countplot(x='binary_target',data=df,palette='YlOrRd')
plt.savefig('eda_files/target_count.png')
```

```python
df.isnull().sum()
```

```python
#Série temporal para verificar as quedas de cada sensor
df.plot(subplots =True, sharex = True, figsize = (30,50))

#Pode se verificar que o sensor 15 não possui nenhum valor
#Vemos também que todos seguem um padrão de datas bem semelhante uns com os outros
#Existem alguns sensores que não parecem seguir alguma tendência e que possuem muito ruído (ex: 38)
```

```python
#Verifica a porcentagem de valores nulos dentro dos dados
round((df.isnull().sum() * 100/ len(df)),2).sort_values(ascending=False)
```

```python
failure = df[df['binary_target'] == 1]

plt.figure(figsize=(20, 5))
plt.plot(failure['50'], linestyle = 'none', marker= 'X', color = 'red', markersize = '3')
plt.plot(df['50'], color="grey", alpha= 0.5)
plt.title('Sensor 50')
plt.savefig('eda_files/sensor50.png')
plt.show()

#Por possuir quase 35% de valores faltantes, o melhor a ser feito é apagar a coluna
```

```python
plt.figure(figsize=(20, 5))
plt.plot(failure['51'], linestyle = 'none', marker= 'X', color = 'red', markersize = '3')
plt.plot(df['51'], color="grey", alpha= 0.5)
plt.title('Sensor 51')
plt.savefig('eda_files/sensor51.png')
plt.show()

#Seguindo uma lógica semelhante à coluna 50, não existem valores que auxiliem na modelagem, portanto também será excluída. 
```

```python
plt.figure(figsize=(20, 5))
plt.plot(failure['15'], linestyle = 'none', marker= 'X', color = 'red', markersize = '3')
plt.plot(df['15'], color="grey", alpha= 0.5)
plt.title('Sensor 15')
plt.savefig('eda_files/sensor15.png')
plt.show()
```

```python
#Excluindo a coluna 15 (que só possui valores nulos), 50 e 51.
eda_df = df.drop(columns=['15', '50', '51'], axis=1)
%store eda_df
```

```python
sensors = df.iloc[:, :49]
sensor_name = sensors.columns

for sensor in sensor_name:
    plt.figure(figsize=(20, 5))
    plt.plot(failure[sensor], linestyle = 'none', color ='red', marker='X', markersize ='5')
    plt.plot(df[sensor], color= 'grey', alpha = 0.5)
    plt.title(sensor)
    plt.show()

#Todos seguem um padrao de falhas e dados faltantes, em que quando há uma falha, os dados são perdidos.
```

```python

# Criamos uma única figura com 2 linhas e 1 coluna
# O figsize aqui define o tamanho total da imagem final
fig, axs = plt.subplots(2, 1, figsize=(20, 10))

# --- PRIMEIRO GRÁFICO (Sensor 14) ---
# Usamos axs[0] para acessar a primeira linha
axs[0].plot(df['14'], color="grey", alpha=0.5, label='Normal')
axs[0].plot(failure['14'], linestyle='none', marker='X', color='red', markersize=3, label='Falha')
axs[0].set_title('Sensor 14')
axs[0].legend() # Opcional: para identificar o que é vermelho/cinza

# --- SEGUNDO GRÁFICO (Sensor 34) ---
# Usamos axs[1] para acessar a segunda linha
axs[1].plot(df['34'], color="grey", alpha=0.5)
axs[1].plot(failure['34'], linestyle='none', marker='X', color='red', markersize=3)
axs[1].set_title('Sensor 34')

# Ajusta o espaçamento para o título não bater no gráfico de baixo
plt.tight_layout()
plt.show()
```
